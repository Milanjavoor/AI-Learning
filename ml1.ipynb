```python

import pandas as si 
import sklearn as sk 
import numpy as np
```

```python
def prepare_country_stats(oecd_bli,gdp_per_capita):
    oecd_bli=oecd_bli[oecd_bli["INEQUALITY"]=="TOT"]
    oecd_bli=oecd_bli.pivot(index="Country",columns="Indicator",values="Value")
    gdp_per_capita=gdp_per_capita.rename(columns={"2015":"GDP per capita"})
    gdp_per_capita=gdp_per_capita.set_index("Country")
    full_stats=si.merge(left=oecd_bli,right=gdp_per_capita,left_index=True,right_index=True)
    full_stats.sort_values(by="GDP per capita",inplace=True)
    remove_indices=[0,1,33,8,34,35]
    indices_kept=list(set(range(36))-set(remove_indices))
    features = [
    "GDP per capita",
    "Employment rate",
    "Educational attainment",
    "Air pollution",
    "Employees working very long hours",
    "Life satisfaction"
    ]
    return full_stats[features].iloc[indices_kept]
```

```python
import urllib
import os
datapath=os.path.join("datasets","lifesat","")
root="https://raw.githubusercontent.com/ageron/handson-ml2/master/"
os.makedirs(datapath,exist_ok=True)
for filename in ("oecd_bli_2015.csv","gdp_per_capita.csv"):
    print("downloadind",filename)
    url=root+"datasets/lifesat/"+filename
    urllib.request.urlretrieve(url,datapath+filename)
```

```python
import pandas as si 
import numpy as np 
import sklearn.linear_model
import matplotlib.pyplot as plt 

oecd_bli=si.read_csv(datapath+"oecd_bli_2015.csv",thousands=",")
gdp_per_capita=si.read_csv(datapath+"gdp_per_capita.csv",thousands=",",delimiter="\t",encoding="latin1",na_values="n/a")
country_stats=prepare_country_stats(oecd_bli,gdp_per_capita)
featurs=["Employment rate",
    "Educational attainment",
    "Air pollution",
    "Employees working very long hours",
    "GDP per capita"]
x=country_stats[featurs]
y=np.c_[country_stats["Life satisfaction"]]

model=sklearn.linear_model.LinearRegression()
model=model.fit(x,y)

emprate=int(input("Enter employment rate"))
edu=int(input("Enter Educational attainment rate"))
emp=int(input("employees working long hours:"))
air=int(input("air pollution level"))
gdp=int(input("Enter GDP per capita of the country"))
new_country=[[emprate,edu,air,emp,gdp]]
country=input("Enter name of the country")
print("The life satisfaction rate for the country",country,"is ",model.predict(new_country))


```
